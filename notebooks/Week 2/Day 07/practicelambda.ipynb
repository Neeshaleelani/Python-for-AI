You have prices = [199, 49, 999, 15, 320]. Use map() with a lambda to apply a 10% discount to every price (multiply by
0.9), then round each result to a whole number using round().

```python
prices = [199, 49, 999, 15, 320]
discount =list(map(lambda p: round(p-p*0.1),prices ))
print(discount)
```

You have names = ["ali", "Sara", "BILAL", "zara"]. In one line, first filter out any name that is not entirely lowercase
(hint: use str.islower()), then map the survivors to their capitalized form (hint: use str.capitalize()). Print the final list.

```python
names = ["ali", "Sara", "BILAL", "zara"]
notcapital = filter(lambda n: n.islower(),names)
capital = list(map(lambda n: n.upper(),notcapital))
print(capital)
```
