# 9. The classic: a word frequency counter

This is the problem that makes dictionaries click. It also comes up in interviews constantly.

````python
text ="I am a Neesha Kumari, Neesha is a AshrieTech student and taking course of Python for AI"
words = text.split()
count = 1
Dic ={}
for word in words:
    if word in Dic:
        Dic[word] =count+1
    else:
        Dic[word]=count
print(Dic)
````

Given names = ["Ali", "Sara", "Ahmed"] and marks = [78, 92, 85], write code that builds and prints a
dictionary pairing each name with their mark at the same position.

````python
names,  marks= ["Ali", "Sara", "Ahmed"] , [78, 92, 85]
dic={}
for index, key in enumerate(zip(names)):
    dic[key]=marks[index]
print(dic)
    
````
