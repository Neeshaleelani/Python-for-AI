for Loop: Iterating a String

```python
string = input("Enter any word")
vowels = "aeiou"
count = 0
consonent = 0
for ch in string.lower():
    if ch in  vowels:
        count +=1
    elif ch.isalpha():
        consonent +=1
    else:
        pass
print(f"The {string} has {count} vowels and {consonent} consonent")
```

Practice: Reverse a String

```python
string = input("Enter any word")
reverseString =""
for ch in string:
    reverseString = ch + reverseString
print(f"{string} After Reversed is: {reverseString}")
```

2. Write a program that finds the factorial of a number using a loop (not recursion).

```python
number =int(input("Enter a number"))
factorial=[]
for i in range(1,number):
    if number % i==0:
        factorial.append(i)

print(f"Factorial of {number} are {factorial}")
```

3. Write a program that finds the GCD of two numbers using a while loop.

```python
number1 =int(input("Enter a number"))
number2 =int(input("Enter a number"))

factors1 = []
factors2 = []

GCDL=[]

for i in range(2,number1):
    if number1 % i == 0:
        factors1.append(i)
    else: 
        pass
for j in range(2,number2):
    if number2 % j == 0:
        factors2.append(j)
    else: 
        pass
print(factors1,factors2)

for value in factors1:
    for value2 in factors2:
        if value==value2:
            GCDL.append(value)
        else:
            pass
GCD= max(GCDL)
print(f"GCD of {number1} and {number2} is {GCD}")
```
