```python
class Shape:

    def area(self):
        return 0


class Rectangle(Shape):
    def __init__(self, l, w):
            self.l = l
            self.w = w
    
    def area(self):
        return self.l * self.w


class Circle(Shape):
    def __init__(self, r):
                self.r = r
                
    def area(self):
        return 3.141 * self.r**2


shapes = [
    Rectangle(5, 2),
    Circle(2),
    Shape()
]

for shape in shapes:
    print(shape.area())
```
