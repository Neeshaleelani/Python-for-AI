```python
def is_prime(n):
    if n > 0 and n<=2:
        return True
    for i in range(2, n):
        if n % i == 0:
            return False
        return True

print(is_prime(4)) # True
print(is_prime(10)) # False

numbers = range(0,20)
primes = [i for i in numbers if is_prime(i)]
print(primes)
```

```python
def factorial(n:int):
    fact=1
    if n==0:
        return 1
    else:
        for i in range(0,n):
            if i!=n:
                fact=fact*(n-i)
        return fact

print(factorial(9))
```
