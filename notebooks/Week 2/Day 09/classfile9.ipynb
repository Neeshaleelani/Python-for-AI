```python
def count_up_to(n):
    i=1
    while i<=n:
        yield i
        i+=1
gen_func= count_up_to(5)
for num in gen_func:
    print(num)
```

```python
gen = count_up_to(3)
print(gen)            # a generator object - no code inside has run yet
print(next(gen))      # runs until the first yield
print(next(gen))      # resumes, runs to the next yield
print(next(gen))
```

```python
def Shout(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}...")
        args = (args[0].upper(),) + args[1:]
        result = func(*args, **kwargs)
        return result.upper()
                
    return wrapper

@Shout
def greet(text):
    return "Hello"+" " + text



print(greet("Neesha"))
```
