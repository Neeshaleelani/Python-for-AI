```python
current_staff = {"name": "Neesha", "role": "manager"}

def manager_only(func):
    def wrapper(*args, **kwargs):
        if current_staff["role"] != "manager":
            print(f"Not Allowed: {current_staff['name']} is not a manager.")
            return None
        return func(*args, **kwargs)
    return wrapper

@manager_only
def issue_refund(orderno, amount):
    print(f"Manager: Refunded Rs {amount} for order {orderno}")

issue_refund("A-11", 100)                 
current_staff["role"] = "Cashier"
current_staff["name"]="Amirita"
issue_refund("A-12", 500)                 
```
