Practice: Find Duplicates in a List

```python
string ="aabbbcdaeef"
check=""
duplicate=""
for s in string:
    if s in check:
        if s not in duplicate:
            duplicate= duplicate + s
    else:
        check= check + s
print(f"After removing duplictes {duplicate} form {string} we have {check}")
```

Given a list of numbers, find the second-largest number without sorting the whole list twice or using sorted()[-2].

```python
#max number in list
list= [7,2,9,2,5,11,48,20]
max=0
for index, l in enumerate(list):
    if l>list[index]:
        max=l
    elif max > list[index]:
        max=max
    else:
        max=list[index]
print(max)
```
