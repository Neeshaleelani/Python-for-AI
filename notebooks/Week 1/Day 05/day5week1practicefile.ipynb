Think & Practice

Given results = ["pass", "fail", "pass", "pass", "fail"], use a dictionary and .get() to count how many "pass" and how many "fail"
results there are.

```python
results = ["pass", "fail", "pass", "pass", "fail"]
count ={}
for r in results:
    count[r] =count.get(r,0) +1
print(count)
```
