Real-life analogy: "If it's raining, take an umbrella. Else, wear
sunglasses."

```python
Answer = input("Is it raining? Enter Yes/ No")
if Answer.lower() == "yes":
    print("Take an umbrella")
elif Answer.lower() == "no":
    print("Wear sunglasses")
else:
    print("Invalid input")
```

Ternary (Conditional) Expressions
A one-line shorthand for if...else. Common in interviews.

```python
num = int(input("Enter a number"))
Answer = "Even" if num % 2==0 else "Odd"
print(f"{num} number is: {Answer}")
```

Largest of Three Numbers

```python
num1,num2, num3 = int(input("enter num 1")), int(input("enter num 2")), int(input("enter num3"))

if num1>=num2 and num1>=num3:
    print(f"{num1} is greater")
elif num2>=num1 and num2>=num3:
    print(f"{num2} is greater ")
elif num3>=num2 and num3>=num1:
    print(f"{num3} is greater")
else :
    print(f"All are equal")
```

Think & Practice (Harder)
A triangle has three sides a, b, and c.
Write a program that checks whether the three sides can form a valid triangle.
Hint: the sum of any two sides must be greater than the third side.

```python
a, b ,c = int(input("Enter side 1")), int(input("Enter side 12")), int(input("Enter side 3"))
if (a+b>c) and (a+c>b) and (b+c>a):
    print("Valid Triangle")
else:
    print("Invalid Triangle")
```
