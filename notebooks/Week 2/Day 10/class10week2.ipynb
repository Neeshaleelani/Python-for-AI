```python
class Student:
    def __init__(self, name, marks):
        self.name= name
        self.marks= marks

    def Avg(self):
        return sum(self.marks)/len(self.marks)

    def Grade(self):
        calculatedAvg= self.Avg()
        if calculatedAvg >=80:
            return "Your grade is: A"
        elif calculatedAvg >=70:
            return "Your grade is: B"
        elif calculatedAvg >=60:
            return "Your grade is: C"
        else:
            return "You are Fail"

s1= Student("Neesha", [85,92,78])

print("Name: ",s1.name, "Avg marks: ", s1.Avg(), s1.Grade())


```

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"Refused: {self.owner} tried to withdraw Rs {amount} but has Rs {self.balance}")
            return
        self.balance -= amount

    def transfer(self, other, amount):
        if self.balance >= amount:
            self.withdraw(amount)     
            other.balance.deposit(amount)
        else:
            print("You do not have balance")


ali = BankAccount("Ali", 1000)
ali.withdraw(5000)            # refused
ali.withdraw(400)             # fine
zara = BankAccount("zara",1000)
ali.transfer(zara,100)
print("Balance:", ali.balance)
```
